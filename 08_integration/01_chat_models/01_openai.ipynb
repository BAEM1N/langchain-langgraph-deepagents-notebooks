# 01.01 OpenAI / Azure OpenAI

OpenAI의 관리형 API는 LangChain에서 가장 먼저 만나는 공급자다. `ChatOpenAI`·`AzureChatOpenAI`·`init_chat_model("openai:…")` 3가지 진입점과 GPT-4.1 이후 도입된 **Responses API**(`use_responses_api=True`) 옵션을 다룬다.

## 학습 목표

- `ChatOpenAI` 직접 인스턴스화 vs `init_chat_model("openai:<model>")` v1 헬퍼 경로를 구분한다
- `AzureChatOpenAI`에서 `azure_endpoint`·`api_version`·`azure_deployment` 세 축 매핑을 이해한다
- Responses API(`use_responses_api=True, store=False, include=["reasoning.encrypted_content"]`)로 데이터 보존 없이 reasoning summary를 받는다
- `model.profile` 속성으로 `max_input_tokens`·`tool_calling`·`reasoning_output` 능력치를 조회한다

## 언제 쓰나

- GPT-4.1 / GPT-4o / o-series 등 OpenAI 최신 모델을 사용하는 모든 에이전트
- Microsoft Entra / Azure 구독에 묶인 엔터프라이즈 배포(Azure OpenAI 리소스)
- **Zero Data Retention** 요구가 있는 환경에서 reasoning trace를 암호화 포맷으로 받아야 할 때

## 01.01.1 환경 설정

`langchain-openai` 패키지와 `.env`의 `OPENAI_API_KEY`(Azure의 경우 `AZURE_OPENAI_*`) 가 필요하다.

```python
# !pip install -U langchain langchain-openai
import os
from dotenv import load_dotenv
load_dotenv(override=True)

assert os.environ.get("OPENAI_API_KEY"), "OPENAI_API_KEY 필요"
```

## 01.01.2 기본 사용 — `ChatOpenAI`

`ChatOpenAI(model="gpt-5.4-mini")` 한 줄로 chat model을 만든다. `invoke()`는 단일 AIMessage를, `stream()`은 토큰 단위 제너레이터를 반환한다.

```python
from langchain_openai import ChatOpenAI

model = ChatOpenAI(model="gpt-5.4-mini", temperature=0)
msg = model.invoke("LangChain v1에서 init_chat_model의 역할을 한 문장으로 설명해줘.")
print(msg.content)
print("usage:", msg.usage_metadata)
```

## 01.01.3 `init_chat_model()` — v1 권장 진입점

`init_chat_model("<provider>:<model>")` 는 프로바이더 문자열 하나로 chat 모델을 교체한다. 구성 가능(`configurable`) 파라미터로 프로덕션 A/B 테스트도 간단해진다.

```python
from langchain.chat_models import init_chat_model

m = init_chat_model("openai:gpt-5.4-mini", temperature=0)
print(type(m).__name__, "—", m.invoke("안녕").content)

# configurable_fields로 런타임 모델 스왑
configurable = init_chat_model("openai:gpt-5.4-mini", configurable_fields=("model",))
out = configurable.invoke("한 단어로 인사해줘.", config={"configurable": {"model": "gpt-5.4-nano"}})
print("swap →", out.content)
```

## 01.01.4 `.profile` 속성으로 능력치 조회

`langchain-core` 1.0+ chat 모델의 `.profile` 속성에서 컨텍스트 길이·tool calling·구조화 출력·reasoning 지원 여부를 런타임에 확인한다. 에이전트 팩토리 코드에서 **"이 모델이 tool calling을 지원하는가"**를 판단할 때 쓴다.

```python
profile = model.profile
print("max_input_tokens :", profile.get("max_input_tokens"))
print("max_output_tokens:", profile.get("max_output_tokens"))
print("tool_calling    :", profile.get("tool_calling"))
print("structured_output:", profile.get("structured_output"))
print("reasoning_output :", profile.get("reasoning_output"))
```

## 01.01.5 Responses API — reasoning + zero retention

GPT-5 / o-series reasoning 모델을 OpenAI Responses 엔드포인트로 호출하려면 `use_responses_api=True`를 준다. `store=False`는 **OpenAI 서버에 응답을 저장하지 않음**(ZDR), `include=["reasoning.encrypted_content"]`는 암호화된 reasoning을 응답에 담아 다음 호출에 다시 넣을 수 있게 한다.

```python
reasoning = ChatOpenAI(
    model="gpt-5-mini",
    use_responses_api=True,
    store=False,
    include=["reasoning.encrypted_content"],
)
r = reasoning.invoke("3단계로 나눠서 17의 제곱근을 추정해줘.")
print("content[:200]:", r.content[:200] if isinstance(r.content, str) else r.content)
# response_metadata 에 reasoning summary가 포함된다
print("reasoning keys:", list((r.response_metadata or {}).keys()))
```

### reasoning 모델 — `reasoning_effort`

GPT-5 계열·o-series 등 신규 reasoning 모델은 `reasoning_effort` 파라미터로 추론 깊이를 조절한다. 값은 `"minimal" | "low" | "medium" | "high"` 중 하나. 높을수록 reasoning token이 더 많이 발생하고 응답 품질·지연이 함께 증가한다. `ChatOpenAI(..., reasoning_effort=...)` 또는 `init_chat_model("openai:<model>", reasoning_effort=...)` 두 경로 모두 동일 키워드로 받는다.

```python
deep = ChatOpenAI(model="gpt-5-mini", reasoning_effort="low")
quick = ChatOpenAI(model="gpt-5-mini", reasoning_effort="minimal")

for label, m in [("low", deep), ("minimal", quick)]:
    r = m.invoke("17의 제곱근을 한 줄로 답해줘.")
    print(label, "→", r.content[:80] if isinstance(r.content, str) else r.content)
    print("  usage:", r.usage_metadata)
```

### `UsageMetadataCallbackHandler` — 토큰 사용량 집계

`langchain_core.callbacks.UsageMetadataCallbackHandler`는 한 번의 `invoke`/`stream` 동안 모든 chat 모델 호출에서 발생한 `usage_metadata`(input·output·total token + reasoning/cache breakdown)를 누적한다. `callbacks=[handler]` 또는 `config={"callbacks": [handler]}`로 주입하고, 호출 후 `handler.usage_metadata` 딕셔너리(모델별 집계)를 읽는다.

```python
from langchain_core.callbacks import UsageMetadataCallbackHandler

handler = UsageMetadataCallbackHandler()
model.invoke("LangChain v1을 한 문장으로.", config={"callbacks": [handler]})
model.invoke("LangGraph를 한 문장으로.",   config={"callbacks": [handler]})

print("모델별 누적 usage:", handler.usage_metadata)
```

### `output_version="v1"` — content blocks 표준 포맷

`ChatOpenAI(..., output_version="v1")`는 응답을 LangChain 1.0 표준의 **content blocks** 리스트로 정규화한다. 각 블록은 `{"type": "text" | "reasoning" | "tool_use" | "image" | ...}` 형태로, 멀티모달·reasoning summary·tool 호출이 한 스키마 안에서 일관되게 노출된다. 공급자별 응답 형식 차이를 흡수하기 위해 v1 에이전트에서 권장하는 기본값.

```python
m_v1 = ChatOpenAI(model="gpt-5.4-mini", output_version="v1", temperature=0)
r = m_v1.invoke("한 줄로 인사해줘.")

# v1에서 content는 dict 블록의 리스트
print("type:", type(r.content).__name__)
print("blocks:", r.content)
# text만 모으는 헬퍼
text = "".join(b["text"] for b in r.content if isinstance(b, dict) and b.get("type") == "text")
print("text:", text)
```

### Prompt caching — `implicit` vs `explicit`

OpenAI는 동일한 prefix를 가진 프롬프트에 대해 **prompt caching**을 적용해 input token 단가를 낮추고 지연을 줄인다. 두 모드:

- **implicit** (기본): 1024 token 이상 prefix가 자동 매칭. 별도 설정 없음 — 같은 시스템 프롬프트/few-shot/문서 prefix를 반복적으로 보내기만 하면 된다.
- **explicit**: `cache_control={"type": "ephemeral"}`(Anthropic 스타일) 또는 Responses API의 `prompt_cache_key`로 명시 지정. 정밀 제어가 필요한 멀티테넌트·KV 분리 시나리오용.

캐시 히트는 응답 `usage_metadata["input_token_details"]["cache_read"]` 에서 확인할 수 있다.

```python
# implicit caching — 동일 prefix 두 번 호출. 두 번째 응답에서 cache_read가 잡힌다.
long_prefix = ("LangChain 도큐먼트:\n" + "샘플 문장. " * 400)  # 1024 token 이상 보장

cached_model = ChatOpenAI(model="gpt-5.4-mini", temperature=0)
r1 = cached_model.invoke(long_prefix + "\n\n질문: 한 단어로 인사.")
r2 = cached_model.invoke(long_prefix + "\n\n질문: 다른 단어로 인사.")

print("1차 usage:", r1.usage_metadata)
print("2차 usage:", r2.usage_metadata)
# input_token_details.cache_read 가 0보다 크면 캐시 히트
print("2차 cache_read:", (r2.usage_metadata or {}).get("input_token_details", {}).get("cache_read"))

# explicit (Responses API) — prompt_cache_key 로 캐시 네임스페이스를 분리
explicit = ChatOpenAI(
    model="gpt-5-mini",
    use_responses_api=True,
    model_kwargs={"prompt_cache_key": "tenant-42:doc-v3"},
)
print("explicit 객체:", type(explicit).__name__)
```

## 01.01.6 `AzureChatOpenAI` — 엔터프라이즈 배포

Azure OpenAI는 **리소스 엔드포인트 + API 버전 + 배포(deployment) 이름** 3축이 필요하다. 환경변수로 넣어 두면 생성자는 생략 가능하다. (이 셀은 자격이 없으면 그냥 건너뛰어도 된다 — 아래 조건 블록은 key 유무만 확인한다.)

```python
from langchain_openai import AzureChatOpenAI

have_azure = all(os.environ.get(k) for k in ["AZURE_OPENAI_ENDPOINT", "AZURE_OPENAI_API_KEY", "AZURE_OPENAI_API_VERSION"])
print("Azure 자격 감지:", have_azure)

if have_azure:
    azure = AzureChatOpenAI(
        azure_deployment=os.environ.get("AZURE_OPENAI_DEPLOYMENT", "gpt-5.4-mini"),
        api_version=os.environ["AZURE_OPENAI_API_VERSION"],
    )
    print("azure →", azure.invoke("ping").content[:50])
else:
    print("자격 없음 — Azure 경로는 스킵. ChatOpenAI 경로만 사용해도 무방.")
```

## 01.01.7 Tool calling + structured output

`bind_tools([...])`로 함수 호출을 활성화하고, `with_structured_output(Schema)`로 Pydantic 모델 구조를 강제한다. OpenAI는 v1 에이전트에서 가장 호환성이 좋은 기본값이다.

```python
from pydantic import BaseModel, Field
from langchain_core.tools import tool

@tool
def get_price(ticker: str) -> float:
    """티커의 가격을 반환한다."""
    return {"AAPL": 210.3, "GOOG": 180.1}.get(ticker.upper(), 0.0)

tool_model = model.bind_tools([get_price])
resp = tool_model.invoke("AAPL 가격 얼마야?")
print("tool_calls:", resp.tool_calls)

class Quote(BaseModel):
    ticker: str = Field(..., description="심볼")
    price: float

structured = model.with_structured_output(Quote)
print("structured →", structured.invoke("GOOG가 180.1달러야. 구조화해줘."))
```
