## 1. What are Deep Agents?

**Deep Agents** is an **agent harness** framework from the LangChain team.
It assembles the infrastructure needed for autonomous agents that work across complex multi-step tasks:

- **Context management** — file tools (`ls`, `read_file`, `write_file`, `edit_file`, `delete`) plus summarization
- **Code execution** — sandboxed execution or constrained local execution
- **JavaScript interpreter** — QuickJS runtime for lightweight deterministic helper logic
- **Pluggable backends** — in-memory state, local disk, LangGraph store, ContextHub, sandbox
- **Subagent delegation** — context isolation and parallel work
- **Task planning (optional)** — add `TodoListMiddleware` to expose `write_todos`
- **Long-term memory** — cross-thread knowledge through LangGraph stores
- **Filesystem permissions** — declarative read/write/delete rules
- **Human-in-the-loop** — approval workflows with `interrupt_on`
- **Reusable skills** — specialized workflows and gated skill tools

Deep Agents builds on LangChain agent components and uses **LangGraph** as the execution engine for durable execution and streaming.

## 2. SDK vs CLI

Deep Agents is available in two forms:

| Category | Deep Agents SDK | Deep Agents CLI |
|------|----------------|----------------|
| **Package** | `deepagents` | `deepagents-cli` |
| **Purpose** | Build agents programmatically | Use a coding agent directly from the terminal |
| **Install** | `pip install deepagents` | `uv tool install deepagents-cli` |
| **Usage** | Call `create_deep_agent()` from Python | Run `deepagents-cli` in the terminal |
| **Customization** | Full API access (tools, backends, middleware) | Config files + slash commands |
| **Best fit** | App integrations and automation pipelines | Interactive coding assistance |

> In this course, we focus primarily on the **SDK**.

## 3. Six Core Concepts

### 3.1 Planning
Planning is opt-in. Add LangChain's `TodoListMiddleware` when the agent should use `write_todos` to break complex work into a structured task list.

### 3.2 Context Management
Deep Agents manages large amounts of information generated during a task:
- **Offloading**: large content can be written to files while only a pointer stays in context
- **Summarization**: conversation history can be compressed as it approaches the model limit

### 3.3 Backends
The agent filesystem is implemented with **pluggable backends**:
- `StateBackend` — stores files in agent state (thread-scoped default)
- `FilesystemBackend` — local disk access (`virtual_mode=True` by default)
- `StoreBackend` — cross-thread persistence with a required namespace
- `CompositeBackend` — route paths to different backends

### 3.4 Subagents
The main agent can delegate specialized work to subagents. Default isolated mode gives each call separate context; experimental `mode="fork"` inherits parent context.

### 3.5 Memory & Skills
Deep Agents supports both always-on memory and on-demand skills:
- **Memory** — reusable context through `AGENTS.md` and stores
- **Skills** — progressive disclosure through `SKILL.md` and optional `metadata.include_tools`

### 3.6 Quality Gates
`RubricMiddleware` uses criterion-level evaluation and grader hooks to inspect output quality and drive runtime revision when needed.

## 4. Comparison with Other Frameworks

| Capability | LangChain Deep Agents | OpenCode | Claude Agent SDK |
|------|----------------------|----------|------------------|
| **Model support** | Model-agnostic (Anthropic, OpenAI, 100+ providers) | 75+ providers, including Ollama | Claude-only |
| **License** | MIT | MIT | MIT (SDK) / proprietary (Claude Code) |
| **SDK** | Python, TypeScript + CLI | Terminal, desktop, IDE | Python, TypeScript |
| **Sandboxing** | Integrated as a tool (Modal, Daytona, etc.) | Not supported | Not supported |
| **Pluggable backends** | O (State, FS, Store, Composite) | X | X |
| **Time travel** | O (via LangGraph) | X | O |
| **Observability** | Native LangSmith support | X | X |
| **Built-in file tools** | O | O | O |
| **Human-in-the-loop** | O | X | O |

Deep Agents is especially strong when you want to build agents that need **planning + files + memory + subagents** in one integrated stack.
